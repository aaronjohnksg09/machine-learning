```python
#Standardize Missing-Value Markers

# ---  Audit: scan for common textual missing markers before touching anything ---
text_markers = {"na", "n/a", "unknown", "not known", "none", "-", "null", "nan", ""}

audit_rows = []
for col in df.select_dtypes(include="object").columns:
    vals = df[col].astype(str).str.strip().str.lower()
    hits = vals.isin(text_markers).sum()
    if hits > 0:
        audit_rows.append((col, hits, sorted(set(vals[vals.isin(text_markers)]))))

audit_df = pd.DataFrame(audit_rows, columns=["column", "marker_count", "markers_found"])
audit_df.sort_values("marker_count", ascending=False)
```

```python
import numpy as np
```

```python
# --- 5.2 Columns where blank/none/unknown = true missing (identity/free-text fields) ---
missing_as_null_cols = [
    "funder", "installer", "wpt_name", "subvillage",
    "scheme_name", "scheme_management"
]

for col in missing_as_null_cols:
    before = df[col].isnull().sum()
    cleaned = df[col].astype(str).str.strip()
    is_marker = cleaned.str.lower().isin(text_markers) | df[col].isnull()
    df[col] = df[col].where(~is_marker, np.nan)
    after = df[col].isnull().sum()
    print(f"{col}: {before} -> {after} nulls")
```

```python
# --- Domain-categorical columns: keep "unknown" as a real category, ---
# --- but blank strings still become true nulls ---
categorical_domain_cols = [
    "water_quality", "quality_group", "quantity", "quantity_group",
    "payment", "payment_type", "source", "source_type", "source_class",
    "waterpoint_type", "waterpoint_type_group",
    "extraction_type", "extraction_type_group", "extraction_type_class",
    "management", "management_group"
]

for col in categorical_domain_cols:
    df[col] = df[col].astype(str).str.strip()
    df[col] = df[col].replace("", np.nan)
    df[col] = df[col].replace("nan", np.nan)  # guard against str(NaN) -> "nan"
```

```python
# ---  Numeric sentinel values acting as disguised nulls ---
# 0 in these fields is not a physically plausible reading -> treat as missing
sentinel_zero_cols = ["gps_height", "longitude", "latitude", "population", "construction_year"]

for col in sentinel_zero_cols:
    zero_count = (df[col] == 0).sum()
    print(f"{col}: {zero_count} zero-value sentinels flagged")

# latitude uses ~-2e-08 as its "zero" due to floating point export quirks
lat_near_zero = df["latitude"].abs() < 1e-6
print(f"latitude near-zero (~-2e-08 placeholder): {lat_near_zero.sum()}")

# amount_tsh == 0 is NOT auto-nulled here — zero payment is a plausible real value
# (free waterpoints exist); flagged for judgment call in Step 12, not Step 5
print(f"\namount_tsh == 0 (left as-is, flagged only): {(df['amount_tsh'] == 0).sum()}")
```

```python
## Step 6: Fix Data Types

# --- Parse dates ---
df["date_recorded"] = pd.to_datetime(df["date_recorded"], format="%Y-%m-%d", errors="coerce")
print("date_recorded dtype:", df["date_recorded"].dtype)
print("Unparseable dates (now NaT):", df["date_recorded"].isnull().sum())
```

```python
# ---  IDs and administrative codes as strings, not numeric ---
# These are identifiers/labels, not quantities — arithmetic on them is meaningless
id_and_code_cols = ["id", "region_code", "district_code"]

for col in id_and_code_cols:
    df[col] = df[col].astype("string")
    print(f"{col}: -> {df[col].dtype}")
```

```python
# ---  True numeric measurement columns ---
numeric_cols = [
    "amount_tsh", "gps_height", "longitude", "latitude",
    "num_private", "population", "construction_year"
]

for col in numeric_cols:
    before_dtype = df[col].dtype
    df[col] = pd.to_numeric(df[col], errors="coerce")
    new_nulls = df[col].isnull().sum()
    print(f"{col}: {before_dtype} -> {df[col].dtype}, nulls after coercion: {new_nulls}")
```

```python
# --- 6.5 Categorical / label columns -> category dtype ---
categorical_cols = [
    "funder", "installer", "wpt_name", "basin", "subvillage", "region",
    "lga", "ward", "recorded_by", "scheme_management",
    "extraction_type", "extraction_type_group", "extraction_type_class",
    "management", "management_group", "payment", "payment_type",
    "water_quality", "quality_group", "quantity", "quantity_group",
    "source", "source_type", "source_class",
    "waterpoint_type", "waterpoint_type_group"
]

for col in categorical_cols:
    df[col] = df[col].astype("category")
```

```python
0## Step 7: Re-check Missingness (post type conversion)

# --- 7.1 Missingness snapshot after Step 6 conversions ---
missing_after_dtype_fix = df.isnull().sum()
missing_pct_after = (missing_after_dtype_fix / len(df) * 100).round(2)

missingness_report = pd.DataFrame({
    "missing_count": missing_after_dtype_fix,
    "missing_pct": missing_pct_after
}).sort_values("missing_count", ascending=False)

missingness_report
```

```python
# --- 7.2 Compare against the raw file's null counts to isolate NEW nulls ---
# df_raw was captured untouched at load time (Step 3)
missing_before = df_raw.isnull().sum()

comparison = pd.DataFrame({
    "missing_before_dtype_fix": missing_before,
    "missing_after_dtype_fix": missing_after_dtype_fix
})
comparison["new_nulls_from_conversion"] = comparison["missing_after_dtype_fix"] - comparison["missing_before_dtype_fix"]

# Only show columns where type conversion actually introduced new nulls
comparison[comparison["new_nulls_from_conversion"] > 0].sort_values(
    "new_nulls_from_conversion", ascending=False
)
```

```python
# ---  Drill into public_meeting / permit: rows where boolean mapping failed ---
for col in ["public_meeting", "permit"]:
    bad_bool_mask = df[col].isnull() & df_raw[col].notnull()
    print(f"{col}: {bad_bool_mask.sum()} unmapped value(s) -> new nulls")
    if bad_bool_mask.sum() > 0:
        print(df_raw.loc[bad_bool_mask, col].unique())
```

```python
# ---  Drill into numeric columns: rows where to_numeric coercion failed ---
numeric_cols = ["amount_tsh", "gps_height", "longitude", "latitude",
                 "num_private", "population", "construction_year"]

for col in numeric_cols:
    bad_numeric_mask = df[col].isnull() & df_raw[col].notnull()
    n_bad = bad_numeric_mask.sum()
    if n_bad > 0:
        print(f"{col}: {n_bad} unparseable value(s)")
        print(df_raw.loc[bad_numeric_mask, col].unique())
```

```python
# ---  Sanity gate: fail loudly if conversion silently nulled a large chunk of a column ---
THRESHOLD_PCT = 5.0  # flag anything that lost more than 5% of its data to bad conversion

flagged = comparison[comparison["new_nulls_from_conversion"] > 0].copy()
flagged["new_null_pct"] = (flagged["new_nulls_from_conversion"] / len(df) * 100).round(2)
risky = flagged[flagged["new_null_pct"] > THRESHOLD_PCT]

if not risky.empty:
    print(" Columns with significant new nulls from type conversion — investigate before proceeding:")
    print(risky)
else:
    print("No column lost more than", THRESHOLD_PCT, "% of its data to type-conversion nulls.")
```

```python
##  Handle Missing Values

# ---  Confirm current missingness landscape before deciding treatment ---
missing_summary = df.isnull().sum()
missing_pct = (missing_summary / len(df) * 100).round(2)
pd.DataFrame({
    "missing_count": missing_summary,
    "missing_pct": missing_pct
}).query("missing_count > 0").sort_values("missing_count", ascending=False)
```

```python
# Drop scheme_name: redundant with scheme_management/management/management_group,
# and ~48% missing on top of that
redundant_cols_to_drop = ["scheme_name"]

df = df.drop(columns=redundant_cols_to_drop)
```

```python
# scheme_management: 6.5% missing, not redundant with `management`
# (similar categories but different counts — distinct signal, worth keeping)
# indicator flag — preserve the fact that it was originally missing
df["scheme_management_was_missing"] = df["scheme_management"].isnull()

# explicit "missing" label rather than leaving NaN
df["scheme_management"] = df["scheme_management"].cat.add_categories(["missing"])
df["scheme_management"] = df["scheme_management"].fillna("missing")
```

```python
# installer: 6.2% true missing, PLUS a disguised-missing sentinel:
# "0" appears 777 times as a literal string, not a real installer name
df["installer"] = df["installer"].replace("0", pd.NA)

df["installer"] = df["installer"].astype("category")

# indicator flag — covers both real NaN and the "0" sentinel folded in above
df["installer_was_missing"] = df["installer"].isnull()

# explicit "missing" label
df["installer"] = df["installer"].cat.add_categories(["missing"])
df["installer"] = df["installer"].fillna("missing")
```

```python
# funder: 6.1% true missing, PLUS the same "0" sentinel seen in installer
# (confirmed: same 777 rows have "0" in both funder and installer —
# likely the same surveyor leaving both blank at once)
df["funder"] = df["funder"].replace("0", pd.NA)

df["funder"] = df["funder"].astype("category")

# indicator flag — covers both real NaN and the "0" sentinel folded in above
df["funder_was_missing"] = df["funder"].isnull()

# explicit "missing" label
df["funder"] = df["funder"].cat.add_categories(["missing"])
df["funder"] = df["funder"].fillna("missing")
```

```python
# wpt_name: Step 5 already converted "none"/"unknown"/etc. to real NaN.
# Step 8's only job here is the indicator flag + explicit "missing" label —
# no marker cleanup needed again.
df["wpt_name"] = df["wpt_name"].astype("category")

df["wpt_name_was_missing"] = df["wpt_name"].isnull()

df["wpt_name"] = df["wpt_name"].cat.add_categories(["missing"])
df["wpt_name"] = df["wpt_name"].fillna("missing")
```
