```python
#Filtering 
# Select homes above a price threshold
ExpensiveHomes=df[df["SalePrice"]> 300000]
print(ExpensiveHomes)
```

```python
##Combine conditions using & (and) and | (or). Parentheses matte##
modern_large_homes = df[
    (df["YearBuilt"] >= 2000) &
    (df["GrLivArea"] > 2000)
]
print(modern_large_homes)
```

```python
##Find homes in selected neighbourhoods##
chosen_areas = df[df["Neighborhood"].isin(["NridgHt", "NoRidge", "StoneBr"])]
print(chosen_areas)
```

```python
##Find missing values in a column
df[df["LotFrontage"].isna()]
```

```python
## Aggregating
#### Use .agg() when you want several summary statistics at once
neighborhood_summary = (
    df.groupby("Neighborhood")
      .agg(
          homes_sold=("SalePrice", "count"),
          average_price=("SalePrice", "mean"),
          median_price=("SalePrice", "median"),
          highest_price=("SalePrice", "max"),
          average_living_area=("GrLivArea", "mean")
      )
      .sort_values("average_price", ascending=False)
)

neighborhood_summary.head(10)
```
